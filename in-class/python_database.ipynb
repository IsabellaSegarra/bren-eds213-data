## Using a cursor

like a tunnel to the data

```python
cur = conn.cursor()

cur.execute("SELECT * FROM Site")
```

```python
# you can use fetchall

cur.fetchall()
```

```python
cur.execute("SELECT * FROM Site")
cur.fetchone() # fetch one row 

cur.fetchone()[0] # fetch just the first column 
```

```python
# We want to interate through the rows and print the site code 
cur.execute("SELECT * FROM Site")

while True:
    row = cur.fetchone()
    if row == None:
        break
    print(f"Site code: {row[0]}")
```

```python
# you can also create table
cur.execute("""CREATE TEMP TABLE temp_nests AS
            SELECT * FROM Bird_nests""")
```

```python
cur.execute("SELECT * FROM temp_nests LIMIT 5").fetchall()
```
